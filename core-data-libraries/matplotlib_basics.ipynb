**❓ What is Matplotlib**

Matplotlib is a Python library for creating visual charts and graphs. It allows you to plot data using lines, bars, histograms, scatter plots, and many other styles.

**⚙️ Why Matplotlib is Important in Machine Learning**

Matplotlib is the **foundation of data visualization** in Python.  
It helps us understand data patterns, relationships, and model performance through visual insights.

**🔍 Why Matplotlib is Essential:**
- 📊 **Data Exploration:** Visualize data distributions, correlations, and trends.
- 🤖 **Model Understanding:** Plot decision boundaries, training curves, feature importance.
- 📈 **Performance Evaluation:** Compare predictions vs actuals, loss curves, accuracy.
- 🧩 **Integration:** Works perfectly with NumPy and Pandas.

> 💡 *If NumPy handles numbers and Pandas handles data, Matplotlib helps you SEE them.*

**Simple Line Plot**

```python
x = np.linspace(0, 10, 50)
y = np.sin(x)

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("x axis")
plt.ylabel("y = sin(x)")
plt.show()
```

**Customize plot**

```python
plt.figure(figsize=(7, 4))
plt.plot(x, y, color="red", linestyle="--", marker="o", label="sin(x)")
plt.title("Customized Line Plot")
plt.xlabel("x values")
plt.ylabel("y values")
plt.legend()
plt.grid(True)
plt.show()
```

**Bar Chart**

```python
students = ["Harry", "Ron", "Hermione", "Draco"]
scores = [88, 75, 96, 83]

plt.bar(students, scores, color="orange")
plt.title("Student Scores - Bar Chart")
plt.xlabel("students")
plt.ylabel("scores")
plt.show()
```

```python
plt.barh(students, scores, color="orange")
plt.title("Horizontal Bar Chart")
plt.ylabel("students")
plt.xlabel("scores")
plt.show()
```

**Scatter Plot**

```python
np.random.seed(42)

x = np.random.randn(50)
y = np.random.randn(50)

print(x[:5])
print(y[:5])
```

```python
plt.scatter(x, y, color="purple", alpha=0.7)
plt.title("Scatter Plot")
plt.xlabel("feature 1")
plt.ylabel("feature 2")
plt.show()
```

**Histogram**

```python
data = np.random.randn(1000)

plt.hist(data, bins=30, color="skyblue", edgecolor="black")
plt.title("Data Distribution (Histogram)")
plt.xlabel("value")
plt.ylabel("frequency")
plt.show()
```

**Pie Chart**

```python
sizes = [40, 30, 25, 5]
labels = ["Python", "Java", "C++", "R"]

plt.pie(sizes, labels=labels, autopct="%1d%%")
plt.title("Programming Languages Popularity")
plt.show()
```

**Multiple Plots (sub-plots)**

```python
x = np.linspace(0, 10, 100)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(x, np.sin(x), "r--")
plt.title("Sine")

plt.subplot(1, 2, 2)
plt.plot(x, np.cos(x), "b--")
plt.title("Cosine")

plt.suptitle("Sine and Cosine Functions")

plt.show()
```

# Quick Reference Guide

**Line Plots**: Primarily used to show the relationship of continuous data i.e. numeric moving across an ordered sequence.

**Bar Charts**: Used to compare the individual values of different categories or separate entities.

**Scatter plots**: Used to check the relationship between features to see if data points group together into clusters.

**Histograms**: Used to check the overall distribution of numeric data, showing if it is uniform, skewed, or contains outliers.

**Pie Charts**: Used to show what percentage of a whole belongs to different parts.
