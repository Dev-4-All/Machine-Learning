**2. While Loop**

```text
while condition is True:
    run this code block repeatedly
```

````python
my_string = "Shubham Kar"

while my_string:
    print(my_string)
    my_string = my_string[0:-1]
````

**3. Loop Controls - break, continue**

````python
# break - to stop the loop prematurely
for i in range(10):
    if i == 3:
        break
    print(i)
````

````python
# continue - to skip the current iteration
for i in range(10):
    if i == 3:
        continue
    print(i)
````
