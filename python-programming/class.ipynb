**1. Define a Basic Class**

```python
class Dog:
    pass
```

**2. Create Class Instance**

```python
dog = Dog()
print(type(dog))
```

**3. Add methods to a class**

Methods are the functions defined inside a class.

```python
class Dog:
    def bark(self): # self means "this instance of the class"
        print("woof!")
```

```python
my_dog = Dog()
my_dog.bark()
```

**4. Use Instance Attributes**

These are the variables associated with an instance of our class. Every object has its own copy of the instance attribute.

```python
class Dog:
    def __init__(self, name, age): # constructor
        self.name = name
        self.age = age

    def bark(self):
        print("woof!")
```

```python
dog_1 = Dog(name="Leo", age=4)
print(dog_1.name)
print(dog_1.age)
dog_1.bark()

dog_2 = Dog(name="Nancy", age=5)
print(dog_2.name)
print(dog_2.age)
dog_2.bark()
```

**5. Inheritance**

It allows a subclass to use the properties and methods of a superclass.

```python
class Puppy(Dog):
    def play(self):
        print(f"{self.name} is playing!")
```

```python
puppy = Puppy(name="Max", age=1)
print(puppy.name)
print(puppy.age)
puppy.bark()
puppy.play()
```
