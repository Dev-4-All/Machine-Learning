**1. Define a simple function**

````python
# define greet function
def greet():
    print("Hello! Welcome to my place.")
````

````python
# call the greet function
greet()
````

**2. Function with parameters**

````python
def greet_user(name):
    print(f"Hello {name}!")
````

````python
greet_user("Sid") # easier for short functions with few arguments
greet_user(name="Angela") # less error prone
````

````python
def add(a, b):
    result = a + b
    print(f"{a} + {b} = {result}")
````

````python
add(10, 90)
````

**3. Return a value**

````python
def square(n):
    return n*n

# store the result in a variable and print it later
result = square(3)
print(f"Square of 3 is {result}")

# print directly
print(f"Square of 10 is {square(10)}")
````

**4. Default Parameters**

````python
def add(a=0, b=0):
    result = a + b
    print(f"{a} + {b} = {result}")
````

````python
add(10, 20)
add(10)
add()
````

**5. Global vs Local variables**

````python
x = 5 # module level scope

def test():
    y = 10 # exists only during the execution of this function
    print(x)
    print(y)

test()
print(x)
print(y)
````
