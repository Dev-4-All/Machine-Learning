**Basic If Statement**

```text
if condition is true
    then execute this code block
```

````python
age = int(input("Enter your age: "))

# Check if the person is 18 or older
if age >= 18:
    print("You are an adult.")
````

**If and Else**

````python
age = int(input("Enter your age: "))

if age >= 18:
    print("You can vote.")
else:
    print("Sorry, you are too young to vote.")
````

**If Elif Else**

elif -> else if

````python
marks = int(input("Enter your marks out of 100: "))

if marks >= 90:
    print("Grade A")
elif marks >= 75: # 75-89
    print("Grade B")
elif marks >= 60: # 60-74
    print("Grade C")
else: # <= 59
    print("Grade D - keep learning")
````

**Nested If Statement**

````python
age = int(input("Enter your age: "))

if age >= 18:
    have_id = input("Do you have an id? (Y/N)") == "Y"

    if have_id:
        print("Entry allowed.")
    else:
        print("You need to have an id.")
else:
    print("You are not allowed. Must be 18 or older.")
````

**Combining conditions**

````python
age = int(input("Enter your age: "))
have_id = input("Do you have an id? (Y/N)") == "Y"


if age >= 18 and have_id:    
    print("Entry allowed.")
else:
    print("You must be 18+ and posses an id.")
````
