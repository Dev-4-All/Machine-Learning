**List - Ordered and Mutable**

````python
# Creating a list
fruits = ["Apple", "Mango", "Banana"]
print(fruits)
````

````python
# Accessing items using index (starts at 0)
print(fruits[0], fruits[-1])
````

````python
# Change item
fruits[1] = "Orange"
print(fruits)
````

````python
# Add an item to a list (end)
fruits.append("Guava")
print(fruits)
````

````python
# Remove an item from a list
fruits.remove("Banana")
print(fruits)
````

````python
# Loop through a list
for fruit in fruits:
    print(fruit)
````

````python
multiple_types = [0, 0.0, "", False]
````

````python
print(type(multiple_types))
````

````python
# Find the number of items in a list
print(len(multiple_types))
````

**Tuple - Ordered and Immutable**

````python
colors = ("red", "green", "blue")
print(colors)
print(type(colors))
````

````python
# Accessing items
print(colors[0], colors[2])
````

````python
# Trying to change an item would cause an error
# colors[1] = "white"
````

````python
for color in colors:
    print(color)
````

````python
a = [1, 2, 3, 4, 5]
b = tuple(a)
print(a, b)
print(type(a), type(b))
````

Set - Unordered, Mutable & Unique

````python
students = {"Alice", "Sam", "Viola"}
print(students)
print(type(students))
````

````python
students.add("David")
print(students)
````

````python
students.remove("Sam")
print(students)
````

````python
# Loop through set
for student in students:
    print(student)
````

````python
duplicates_list = [1, 1, 2, 2, 3, 3]
duplicates_set = {1, 1, 2, 2, 3, 3}

print(duplicates_list)
print(duplicates_set)
````

````python
# Not recommended
temp = [8, 9, 1, 4, 2, 2, 3, 7]
temp_set = set(temp)
my_list = list(temp_set)
print(my_list)
````

**Dictionary - Unordered, Mutable & holds Key-Value pairs**

Keys must be unique!

````python
person = {
    "name": "Shubham",
    "age": 22,
    "is_student": True
}

print(person)
````

````python
# Updation
person["name"] = "Sam"
````

````python
person["age"]
````

````python
print(type(person))
````

````python
# Adding a new key-value pair
person["lives_in"] = "London"

print(person)
````

````python
# Loop through dictionary
for key, value in person.items():
    print(key, "->", value)
````

### ✅ Python Data Structures Comparison

| **Type**      | **Ordered**     | **Changeable** | **Allows Duplicates** | **Indexable**       |
|---------------|------------------|----------------|------------------------|----------------------|
| **List**       | ✅ Yes            | ✅ Yes         | ✅ Yes                 | ✅ Yes               |
| **Tuple**      | ✅ Yes            | ❌ No          | ✅ Yes                 | ✅ Yes               |
| **Set**        | ❌ No             | ✅ Yes         | ❌ No                  | ❌ No                |
| **Dictionary** | ✅ Yes *(by key)* | ✅ Yes         | ❌ No *(keys only)*    | ✅ Yes *(by key)*    |
